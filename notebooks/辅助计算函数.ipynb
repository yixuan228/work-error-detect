# 整合各种辅助计算函数

### 计算各阶段及全程料肉比

```python
import pandas as pd
from feed_analysis.config.path import PATH_GROWTH

def cal_day_weight(df, start_age, end_age) -> float:
    """
    计算指定日增重，单位g
    """

    df = df[(df['age'] >= start_age) & (df['age'] <= end_age)].copy()
    if df.empty:
        return None
    else:
        # 增重 = 当前阶段末体重 - 当前阶段初体重
        weight_gain = df['weight_fit'].iloc[-1] - df['weight_fit'].iloc[0]
        # 天数 = 当前阶段末年龄 - 当前阶段初年龄
        days = df['age'].iloc[-1] - df['age'].iloc[0]
        # 日增重
        adg = weight_gain / days *1000 if days != 0 else None
        return adg
```

```python
# 使用示例
df = pd.read_excel(PATH_GROWTH / '汇兴4栋生长数据-poly2-log拟合(延长).xlsx')
phase_adg_df = cal_day_weight(df, 71, 156)
print(phase_adg_df)
```
